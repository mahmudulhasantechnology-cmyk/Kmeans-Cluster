```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv("mall customers.csv")
df.head()
```

```python
df.columns
```

```python
df = df.rename(columns={"Spending Score (1-100)": "score", "Annual Income (k$)":"income"})
```

```python
df.head()
```

```python
df["Gender"]= df["Gender"].map({"Male":1, "Female":0})
```

```python
df.head()
```

```python
sns.pairplot(df)
```

```python
kmean= KMeans()
```

```python
model = kmean.fit(df[["income", "score"]])
```

```python
cernters = model.cluster_centers_
```

```python
cernters[:,0]
```

```python
cernters
```

```python
cernters[:,1]
```

```python
pred = model.predict(df[["income", "score"]])
```

```python
pred
```

```python
df["pred_clu"]= pred
```

```python
df.head()
```

```python
df0= df[df["pred_clu"]==0]
df1= df[df["pred_clu"]==1]
df2= df[df["pred_clu"]==2]
df3= df[df["pred_clu"]==3]
df4= df[df["pred_clu"]==4]
df5= df[df["pred_clu"]==5]
df6= df[df["pred_clu"]==6]
df7= df[df["pred_clu"]==7]
```

```python
df1
```

```python
income = 60
score = 50
pred = model.predict([[income, score]])
pred
```

```python
plt.scatter(df0["income"], df0["score"],label="cluter 0")
plt.scatter(df1["income"], df1["score"],label="cluter 1")
plt.scatter(df3["income"], df3["score"],label="cluter 3")
plt.scatter(df2["income"], df2["score"],label="cluter 2")
plt.scatter(df4["income"], df4["score"],label="cluter 4")
plt.scatter(df5["income"], df5["score"],label="cluter 5")
plt.scatter(df6["income"], df6["score"],label="cluter 6")
plt.scatter(df7["income"], df7["score"],label="cluter 7")

plt.scatter(cernters[:,0], cernters[:,1],marker="*", s=100,label="cluter center")
plt.scatter(income, score, color ="black", label= "test_data")
plt.legend()
```

```python
# elbbow method
wcss= []

clsr = range(2,20)
for k in clsr:
    kmean1 = KMeans(n_clusters=k)
    model1 = kmean1.fit(df[["income", "score"]])
    wcss.append(model1.inertia_)
```

```python
wcss
```

```python
plt.figure(figsize=(4,4))
plt.plot(clsr,wcss, marker="o", color="r")
```

```python
# pip install kneed
```

```python
from kneed import KneeLocator
knl = KneeLocator(clsr, wcss, curve= "convex",direction="decreasing")
```

```python
knl.plot_knee()
```

```python
# with 5 cluster
new_km= KMeans(n_clusters=5)
model_final = new_km.fit(df[["score","income"]])
```

```python
new_cen= model_final.cluster_centers_
```

```python
pred_final = model_final.predict(df[["score","income"]])
```

```python
pred_final
```

```python
df["pred_finnal"]= pred_final
df.head()
```

```python
df0= df[df["pred_finnal"]==0]
df1= df[df["pred_finnal"]==1]
df2= df[df["pred_finnal"]==2]
df3= df[df["pred_finnal"]==3]
df4= df[df["pred_finnal"]==4]
```

```python
plt.scatter(df0["income"], df0["score"],label="cluter 0")
plt.scatter(df1["income"], df1["score"],label="cluter 1")
plt.scatter(df3["income"], df3["score"],label="cluter 3")
plt.scatter(df2["income"], df2["score"],label="cluter 2")
plt.scatter(df4["income"], df4["score"],label="cluter 4")


plt.scatter(new_cen[:,0], new_cen[:,1],marker="*", s=100,label="cluter center")
plt.legend()
```
